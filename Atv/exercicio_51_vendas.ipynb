# Análise de Vendas Diárias e Mensais

Exercício 51 - Ciência de Dados

```python
import pandas as pd
import matplotlib.pyplot as plt

```

```python
# Carregar os dados
df = pd.read_excel('dados_vendas.xlsx')

# Visualizar as primeiras linhas
df.head()
```

```python
# Converter coluna de data (ajuste o nome se necessário)
df['Data'] = pd.to_datetime(df['Data'])

# Ordenar por data
df = df.sort_values('Data')
```

```python
# Gráfico de vendas diárias
plt.figure()
plt.plot(df['Data'], df['Quantidade Vendida'])
plt.title('Vendas Diárias')
plt.xlabel('Data')
plt.ylabel('Quantidade de Vendas')
plt.xticks(rotation=45)
plt.show()
```

```python
# Dia com maior venda
max_row = df.loc[df['Quantidade Vendida'].idxmax()]
print("Maior venda:", max_row['Quantidade Vendida'], "no dia", max_row['Data'])

# Dia com menor venda
min_row = df.loc[df['Quantidade Vendida'].idxmin()]
print("Menor venda:", min_row['Quantidade Vendida'], "no dia", min_row['Data'])
```

```python
# Média de vendas
media = df['Quantidade Vendida'].mean()
print("Média de vendas:", media)
```

```python
# Criar coluna de mês
df['mes'] = df['Data'].dt.month

# Média mensal
media_mensal = df.groupby('mes')['Quantidade Vendida'].mean()
media_mensal
```

```python
# Gráfico de barras - média mensal
plt.figure()
media_mensal.plot(kind='bar')
plt.title('Média de Vendas Mensais')
plt.xlabel('Mês')
plt.ylabel('Média de Vendas')
plt.show()
```

## Explicação da Análise de média de vendas mensais

Ao analisar a média de vendas mensais, percebe-se que não existem grandes variações ao longo do tempo. Os resultados costumam se manter entre 105 e 110 vendas por mês, mostrando uma constância no desempenho. Isso acontece porque, quando em algum dia as vendas ficam abaixo do esperado, normalmente nos dias seguintes ocorre uma recuperação natural. Dessa forma, essa compensação ajuda a equilibrar os números e mantém a média mensal estável em relação aos outros meses.
